ALLUVO-Recommendation system<br>
<br>
Alluvo Recommendation System is a personalized recommendation engine designed for a brand-centric reels platform, where users discover products through short-form videos. The system analyzes user interactions such as views, likes, comments, shares, watch time, and brand preferences to generate a ranked feed of relevant reels tailored to each user’s interests. It leverages a hybrid approach combining content-based and collaborative filtering techniques to enhance engagement, improve product discovery, and increase conversion rates while handling cold-start scenarios for new users and brands.

# Selected Datasets for Recommendation System (tasneem)


 1️⃣ RetailRocket – E-commerce Behavior Dataset  
Link: https://www.kaggle.com/datasets/retailrocket/ecommerce-dataset  

**Description:**  
A real-world e-commerce dataset containing user interactions such as views, add-to-cart, and transactions.

**Key Features:**  
- Implicit feedback  
- Session-based events  
- Item properties  
- Category hierarchy  

**Why Selected:**  
Suitable for building collaborative, session-based, and deep recommendation models for e-commerce systems.
